**To build a multiple linear regression model with StatsModels.**

Using pandas and statsmodel

```python
#1 Import the required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from scipy import stats
from statsmodels.formula.api import ols
```

```python
#import dataset
from google.colab import files

uploaded = files.upload()
```

```python
#load dataset
df = pd.read_csv("logistics_routes.csv")
```

```python
#dataset inspection
df.head()
```

```python
print(df.columns)
```

```python
#defining dependent and independent vars
X = df[['Distance x_i1 (tens of km)',
        'Traffic Index x_i2 (1-10)',
        'Peak Window x_i3 (0 or 1)']]

y = df['Delay Time y_i (minutes)']
```

```python
#fit model
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())
```

```python
#print statistics
print("Coefficients:")
print(model.params)

print("SSE:", model.ssr)

print("Error Variance:", model.mse_resid)

print("Standard Errors:")
print(model.bse)

print("T-statistics:")
print(model.tvalues)

print("P-values:")
print(model.pvalues)
```

```python
#new predictions
new_route = pd.DataFrame({
    'const': [1],
    'Distance x_i1 (tens of km)': [26],
    'Traffic Index x_i2 (1-10)': [4],
    'Peak Window x_i3 (0 or 1)': [1]
})

prediction = model.predict(new_route)

print("Predicted Delivery Delay:", prediction.iloc[0])
```

```python
# Confidence interval for mean and prediction interval for one shipment
prediction_results = model.get_prediction(new_route)
print(prediction_results.summary_frame(alpha=0.05))
```

# Using **numpy**

```python
x1 = np.array([10, 14, 19, 24, 28, 33, 38, 42])
x2 = np.array([2, 3, 2, 5, 4, 6, 5, 7])
x3 = np.array([0, 1, 0, 1, 0, 1, 0, 1])
y = np.array([44, 58, 61, 79, 82, 99, 101, 118])
```

```python
X = np.column_stack((np.ones(8), x1, x2, x3))
```

```python
XtX = X.T @ X
Xty = X.T @ y
beta = np.linalg.solve(XtX, Xty)
```

```python
# Fitted values, residuals and error variance
y_hat = X @ beta
residuals = y - y_hat
SSE = np.sum(residuals ** 2)

n = len(y)
p = X.shape[1]
df = n - p
s2 = SSE / df
```

```python
# Coefficient standard errors and t statistics
cov_beta = s2 * np.linalg.inv(XtX)
standard_errors = np.sqrt(np.diag(cov_beta))
t_values = beta / standard_errors

print("X transpose X:")
print(XtX)
print("X transpose y:", Xty)
print("Coefficients:", beta)
print("SSE:", SSE)
print("Residual variance:", s2)
print("Standard errors:", standard_errors)
print("t statistics:", t_values)
```

```python

# Prediction for the new route
x0 = np.array([1, 26, 4, 1])
prediction = x0 @ beta
print("Predicted delay:", prediction)
```
